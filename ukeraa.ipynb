```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
main_url = 'https://www.ukeraa.com/'
```

```python
main_page = requests.get(main_url)
main_page
soup = BeautifulSoup(main_page.content,'html')
```

```python
header = (soup.find('ul', class_ = 'navbar__menu').text.strip()
          .replace('उकेरा सिरिज','').replace('उकेरा खोज','').replace('उकेरा स्पेशल','')).strip()
print(header)





```

```python
url_home = 'https://www.ukeraa.com/news/detail/171964/'
```

```python
page1 = requests.get(url_home)
page1
soup1 = BeautifulSoup(page1.content,'html')
```

```python
header1 = soup1.find('h1', class_ = 'desc__title').text.strip()
print(header1)
```

```python
body1 = (soup1.find('div', id = 'newsbodylateradd').text.strip()
         .replace('काठमाडौं :','')).strip()
print(body1)
```

```python
article1 = soup1.find_all('img')
image1 = article1[3]['src']
print(image1)
```
